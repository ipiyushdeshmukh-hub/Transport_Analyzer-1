```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Load data
vehicles = pd.read_csv("vehicles.csv")
trips = pd.read_csv("trips.csv")

print("Vehicles Data:")
display(vehicles)

print("Trips Data:")
display(trips)

# ----------------------------------------
# VEHICLE INSIGHTS
# ----------------------------------------

# Most fuel efficient
top_fuel = vehicles.sort_values(by="FuelEfficiency", ascending=False)[["Vehicle", "FuelEfficiency"]]
print("Most Fuel Efficient Vehicles:")
display(top_fuel)

# Highest capacity
top_capacity = vehicles.sort_values(by="Capacity", ascending=False)[["Vehicle", "Capacity"]]
print("Highest Capacity Vehicles:")
display(top_capacity)

# ----------------------------------------
# TRIP ANALYSIS
# ----------------------------------------

# Total distance traveled by each vehicle
distance_stats = trips.groupby("Vehicle")["Distance"].sum()
print("Total Distance Traveled:")
display(distance_stats)

# Fuel consumption per vehicle
fuel_stats = trips.groupby("Vehicle")["FuelUsed"].sum()
print("Total Fuel Consumption:")
display(fuel_stats)

# Number of trips per vehicle
trip_count = trips["Vehicle"].value_counts()
print("Number of Trips by Each Vehicle:")
display(trip_count)

# ----------------------------------------
# VISUALIZATIONS
# ----------------------------------------

# Plot 1 — Distance traveled
plt.figure(figsize=(8,5))
distance_stats.plot(kind="bar")
plt.title("Total Distance Traveled by Vehicles")
plt.xlabel("Vehicle")
plt.ylabel("Distance (km)")
plt.show()

# Plot 2 — Fuel used
plt.figure(figsize=(8,5))
fuel_stats.plot(kind="bar")
plt.title("Total Fuel Consumption by Vehicles")
plt.xlabel("Vehicle")
plt.ylabel("Fuel Used (liters)")
plt.show()

# Plot 3 — Number of trips
plt.figure(figsize=(8,5))
trip_count.plot(kind="bar")
plt.title("Number of Trips per Vehicle")
plt.xlabel("Vehicle")
plt.ylabel("Trips")
plt.show()
```
